```python
## PRIMERO IMPORTAR LA LIBRERIA REQUESTS, LUEGO INICIO LA CONEXION CON LA API ##
```

```python
import requests
base_url = "https://api-colombia.com/"
endpoint = "/api/v1/Airport"
requests.get(base_url + endpoint)

```

```python
## AL OBTENER UN RESPUESTA DE CODIGO 200 SE ENTIENDE QUE LA CONEXION ES EXITOSA ##
```

```python
respuesta_api = requests.get(base_url + endpoint)
```

```python
## AHORA EXTRAIGO LA INFORMACION DE LA API ##
```

```python
respuesta_api.json()
```

```python
repusesta_json = respuesta_api.json()
```

```python
## IMPORTO LA LIBRERIA PANDAS PARA CREAR UN DATAFRAME Y ASI PODER TENER UNA MEJOR VISUALIZACION DE LOS DATOS ##
```

```python
import pandas as pd

df_api = pd.DataFrame.from_dict(repusesta_json)
```

```python
df_api
```

```python
df_api['department'] = df_api['department'].apply(
    lambda x: x.get('name') if isinstance(x, dict) else x
)

df_api['city'] = df_api['city'].apply(
    lambda x: x.get('name') if isinstance(x, dict) else x
)
```
